```python
import os

from dotenv import load_dotenv
from langchain_groq import ChatGroq
from langgraph.graph import MessagesState

load_dotenv()


class pipelinestate(MessagesState):
    raw_input: str = "helo this is ai video"
    edited_text: str = ""
    script_text: str = ""
    final_output: str = ""


llm = ChatGroq(model="llama-3.3-70b-versatile", temperature=0.7)


def editer_node(state: pipelinestate) -> dict:
    if not getattr(state, "raw_input", ""):
        return {"edited_text": ""}

    prompt = f"""You are a helpful writing editor.
Rewrite the following text to be clearer, more polished, and grammatically correct.
Return only the rewritten text.

Text:
{state.raw_input}
"""

    try:
        response = llm.invoke(prompt)
        edited_text = response.content.strip() if hasattr(response, "content") else str(response)
    except Exception as e:
        edited_text = f"Fallback edit: {state.raw_input}"

    return {"edited_text": edited_text}


sample_state = pipelinestate(raw_input="Hello, this is an AI video")
result = editer_node(sample_state)
print(result)
result
```
